# 05 · Evidence graph

The knowledge graph contains distinct plant occurrence, secondary-use report, measured activity and model-only evidence, plus a disconnected CPIC/ClinVar PGx component. Edges carry provenance. Graph connectivity is not causality.

```python
import networkx as nx
g=nx.read_graphml('data/processed/knowledge_graph.graphml')
pgx={n for n in g if str(n).startswith('pgx:')}
plant={n for n in g if str(n).startswith(('plant:','compound:','target:','use:','task:'))}
{'nodes':len(g),'edges':g.number_of_edges(),'pgx_nodes':len(pgx),'phytochemical_nodes':len(plant),'cross_component_edges':[(u,v) for u,v in g.edges if (u in pgx) != (v in pgx)],'sample_edges':list(g.edges(data=True))[:8]}
```
