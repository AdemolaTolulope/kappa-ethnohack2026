# 04 · Verified pharmacogenomic evidence demonstration

This is a separate CYP2C19–clopidogrel CPIC guideline example, not a curcumin/PTGS2 relationship. CPIC recommendations are filtered to the intended CYP2C19 intermediate/poor metabolizer phenotypes and cached `CVI ACS PCI ` population context. ClinVar supplies a molecular identity cross-reference only. Its cached germline classification ('Likely benign; other', not-specified trait) is a disease-association label, not a drug-response call. It is not used to contradict CPIC's 'no function' allele assignment. No patient genotype is represented.

```python
import json
from pathlib import Path
p=json.loads(Path('data/raw/cpic_cyp2c19_clopidogrel.json').read_text())
rs=p['resources']
recs=[x for x in rs['recommendation'] if x.get('population')=='CVI ACS PCI ' and x.get('phenotypes',{}).get('CYP2C19') in ('Intermediate Metabolizer','Poor Metabolizer')]
cv=json.loads(Path('data/raw/clinvar_cyp2c19_star2.json').read_text())
{'gene':rs['gene'][0], 'allele':rs['allele'][0], 'drug':rs['drug'][0], 'pair':rs['pair'][0], 'guideline':rs['guideline'][0], 'recommendations':recs, 'clinvar_identity':{'accession':cv['record']['accession'],'title':cv['record']['title'],'germline_classification':cv['record'].get('germline_classification')}, 'scope_note':cv['scope_note']}
```
